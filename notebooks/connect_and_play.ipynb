Connects to the DMD and writes a test pattern so we can verify our ability to display patterns and the conjugation of our camera to the DMD. 

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Connect to the camera 

import cv2

camera = cv2.VideoCapture(0)

# Disable auto-exposure
camera.set(cv2.CAP_PROP_AUTO_EXPOSURE, 0.25)

# Set manual exposure time (unsure about the units)
camera.set(cv2.CAP_PROP_EXPOSURE, -6)
```

```python
# take an image with the camera


while True:
  # Read a frame from the camera
  ret, frame = camera.read()

  # Break the loop if the frame is not read properly
  if not ret:
    print('Failed to grab frame.')
    break

  # Display the resulting frame
  cv2.imshow('Webcam Stream', frame)

  # Press 'q' on the keyboard to exit the loop
  if cv2.waitKey(1) & 0xFF == ord('q'):
    break

# Release the camera and close all OpenCV windows
camera.release()
cv2.destroyAllWindows()

# ret, frame = camera.read()
# cv2.imshow('ELP Camera', frame)
# if cv2.waitKey(1) & 0xFF == ord('q'):
#     break

# camera.release()
# cv2.destroyAllWindows()
```

```python
from Pycrafter6500 import pycrafter6500 as pyc


pattern = np.zeros((1080, 1920), dtype=np.uint8)
pattern[:, :960] = 1  # Set the left half to 1 

images = [pattern]

# Instantiate the DMD
dlp = pyc.dmd()
dlp.stopsequence()
dlp.changemode(3) # Pattern Sequence Mode

exposure = [1000000] # 1 second
dark_time = [0]
trigger_in = [False]
trigger_out = [1]

dlp.defsequence(images, exposure, trigger_in, dark_time, trigger_out, 0)
dlp.startsequence()

```

```python
# stop the sequence

dlp.stopsequence()
```
